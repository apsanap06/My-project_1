```python
# price is the target variable
x=df.drop('price',axis=1)
y=df['price']
```

```python
print("shape of x",x.shape)
print("shape of y",y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train ,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=200)
```

```python
print("shape of x_train",x_train.shape)
print("shape of x_test",x_test.shape)
```

```python
print("shape of y_train",y_train.shape)
print("shape of y_test",y_test.shape)
```

```python
# FEATURE SCALING
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
sc.fit(x_train)
x_train=sc.transform(x_train)
x_test=sc.transform(x_test)
# we apply scaling on input data only
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
#PREDECTION OF TEST AND TRAINING SET RESULT
y_pred=lr.predict(x_train)
```

```python
lr.coef_
```

```python
# intercept
lr.intercept_
```

```python
# score
lr.score(x_test,y_test)
```

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error,r2_score
```

```python
y_pred=lr.predict(x_test)
y_pred
```

```python
# MAE
print("MAE Value:",mean_absolute_error(y_test,y_pred))
```

```python
#MSE
print("MSE Value:",mean_squared_error(y_test,y_pred))
```

```python
# RMSE
print("RSME Value:",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)
```
